```python
# ============================================================
# GLOBAL SETUP — Run this cell before any lab exercise
# ============================================================
import os, json, time, datetime
from IPython.display import Markdown, display
from openai import OpenAI

# Replace with your actual OpenAI API key
API_KEY = "sk-your-api-key-here"
client = OpenAI(api_key=API_KEY)

def ask(prompt, model='gpt-4o-mini', temperature=0.7, max_tokens=800, system=None):
    """Send a prompt to OpenAI and return the response text."""
    messages = []
    if system:
        messages.append({'role': 'system', 'content': system})
    messages.append({'role': 'user', 'content': prompt})
    r = client.chat.completions.create(
        model=model, messages=messages,
        temperature=temperature, max_tokens=max_tokens
    )
    return r.choices[0].message.content

def show(text, title=None):
    """Display markdown-formatted text with an optional title."""
    if title:
        display(Markdown(f'### {title}'))
    display(Markdown(text))

def ask_with_usage(prompt, model='gpt-4o-mini', temperature=0.7, max_tokens=800, system=None):
    """Send a prompt and return both the response text and token usage."""
    messages = []
    if system:
        messages.append({'role': 'system', 'content': system})
    messages.append({'role': 'user', 'content': prompt})
    r = client.chat.completions.create(
        model=model, messages=messages,
        temperature=temperature, max_tokens=max_tokens
    )
    return r.choices[0].message.content, r.usage

print('Setup complete. All helper functions loaded.')
```

# SESSION 3 — Large Language Models: The Brain Behind AI Tools
## Labs: Tokens & Cost, Context Windows, Model Comparison

### Lab 3.1 — Token Counting and Cost Estimation
**Task:** Understand how tokens work and calculate API costs.

```python
# Lab 3.1 — Token Usage and Cost
prompts_to_test = [
    ('Short prompt', 'What is AI?'),
    ('Medium prompt', 'Explain how large language models work. Cover: training data, '
     'tokenization, attention mechanism, and inference. Under 200 words.'),
    ('Long prompt', 'You are a senior technology consultant advising the board of a '
     '500-person manufacturing company. They are considering investing in AI for '
     'quality control, supply chain optimization, and predictive maintenance. '
     'Write a 300-word executive briefing covering: (1) which use case to prioritize '
     'and why, (2) estimated ROI timeline, (3) key risks, (4) recommended next step. '
     'Professional tone. Specific to manufacturing.'),
]

print(f'{"Type":<16} {"Prompt Tokens":<15} {"Completion Tokens":<18} {"Total":<10} {"Est. Cost (USD)"}')
print('-' * 80)

for label, prompt in prompts_to_test:
    response, usage = ask_with_usage(prompt, temperature=0.3)
    # GPT-4o-mini pricing: $0.15/1M input, $0.60/1M output
    cost = (usage.prompt_tokens * 0.15 + usage.completion_tokens * 0.60) / 1_000_000
    print(f'{label:<16} {usage.prompt_tokens:<15} {usage.completion_tokens:<18} {usage.total_tokens:<10} ${cost:.6f}')

print('\nINSIGHT: Output tokens cost 4x more than input tokens for GPT-4o-mini.')
print('Setting max_tokens wisely controls both cost and output length.')
```

### Lab 3.2 — Context Window and Multi-Turn Conversations
**Task:** Demonstrate how context window enables coherent multi-turn dialogue.

```python
# Lab 3.2 — Multi-Turn Conversation with Context Window
conversation = [
    {'role': 'system', 'content': 'You are a concise AI tutor. Keep replies under 80 words.'},
]

turns = [
    'What is a token in LLMs?',
    'How does that relate to context windows?',
    'What happens when a conversation exceeds the context window?',
    'Summarize everything we discussed in 2 sentences.'
]

total_tokens = 0
for i, user_msg in enumerate(turns, 1):
    conversation.append({'role': 'user', 'content': user_msg})
    r = client.chat.completions.create(
        model='gpt-4o-mini', messages=conversation,
        temperature=0.5, max_tokens=150
    )
    reply = r.choices[0].message.content
    total_tokens += r.usage.total_tokens
    conversation.append({'role': 'assistant', 'content': reply})
    
    print(f'Turn {i} (User): {user_msg}')
    print(f'Turn {i} (AI):   {reply}')
    print(f'  [Tokens this call: {r.usage.total_tokens}]')
    print()

print(f'Total tokens across all turns: {total_tokens}')
print('\nKEY INSIGHT: Each turn sends the ENTIRE conversation history.')
print('Token usage grows with each turn — this is the context window in action.')
print(f'Messages in conversation: {len(conversation)}')
```
