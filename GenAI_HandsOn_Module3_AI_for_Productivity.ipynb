```python
# ============================================================
# GLOBAL SETUP
# ============================================================
import os, json, time, datetime
from IPython.display import Markdown, display
from openai import OpenAI

API_KEY = "sk-your-api-key-here"
client = OpenAI(api_key=API_KEY)

def ask(prompt, model='gpt-4o-mini', temperature=0.7, max_tokens=800, system=None):
    messages = []
    if system: messages.append({'role': 'system', 'content': system})
    messages.append({'role': 'user', 'content': prompt})
    r = client.chat.completions.create(model=model, messages=messages, temperature=temperature, max_tokens=max_tokens)
    return r.choices[0].message.content

def show(text, title=None):
    if title: display(Markdown(f'### {title}'))
    display(Markdown(text))

print('Setup complete.')
```

# SESSION 11 — AI for Professional Email & Communication
## Labs: EMAIL Formula, Email Types, Voice Calibration

### Lab 11.1 — The EMAIL Formula
**Task:** Use the EMAIL formula (Event, Message, Action, Impression, Length) to generate professional emails.

```python
# Lab 11.1 — EMAIL Formula for Professional Emails
def email_prompt(event, message, action, impression, length, tone='Professional'):
    return (
        f'Write a professional email.\n\n'
        f'EVENT (why are you writing): {event}\n'
        f'MESSAGE (core point): {message}\n'
        f'ACTION (what recipient should do): {action}\n'
        f'IMPRESSION (how you want to come across): {impression}\n'
        f'LENGTH: {length}\n'
        f'TONE: {tone}\n\n'
        f'Include: Subject line + email body. Use [NAME] for placeholders.'
    )

# Scenario 1: Deadline extension request
p1 = email_prompt(
    event='Project deliverable due Friday. Two team members are ill.',
    message='Requesting a 1-week extension due to reduced team capacity.',
    action='Approve the extension and confirm new deadline.',
    impression='Responsible, proactive, solution-oriented — not making excuses.',
    length='Under 120 words'
)
show(ask(p1, temperature=0.5), 'Email 1 — Deadline Extension Request')

print('\n' + '=' * 60 + '\n')

# Scenario 2: Bad news to client
p2 = email_prompt(
    event='Software delivery will be delayed by 2 weeks due to a critical vendor dependency.',
    message='Informing client of delay, explaining cause, and sharing revised timeline.',
    action='Acknowledge revised timeline and schedule a sync call.',
    impression='Accountable, transparent, already-acting — not defensive.',
    length='Under 150 words',
    tone='Empathetic but confident'
)
show(ask(p2, temperature=0.5), 'Email 2 — Bad News to Client')

print('\n' + '=' * 60 + '\n')

# Scenario 3: Cold introduction
p3 = email_prompt(
    event='Reaching out to VP Sales at a potential partner company after meeting at a conference.',
    message='Proposing a 20-minute call to explore partnership opportunities.',
    action='Reply with availability for a 20-min call next week.',
    impression='Confident, value-first, respectful of their time.',
    length='Under 100 words',
    tone='Warm but professional'
)
show(ask(p3, temperature=0.5), 'Email 3 — Cold Introduction')
```

### Lab 11.2 — Email Voice Calibration
**Task:** Generate the same email in 3 different tones to understand voice control.

```python
# Lab 11.2 — Same Email, 3 Tones
base_context = (
    'You are a team lead. A team member missed 2 deadlines this month. '
    'You need to address it without being punitive — goal is improvement.'
)
task = 'Write a short email addressing the missed deadlines and setting expectations. Under 120 words.'

tones = {
    'Direct & Firm': 'Direct, clear, firm but fair. No softening language. State the facts and expectations.',
    'Supportive & Coaching': 'Supportive, coaching-oriented. Express concern, offer help, ask what\'s blocking them.',
    'Formal & HR-Appropriate': 'Formal, documented, HR-appropriate. Could be placed in an employee file. Objective language only.',
}

for tone_name, tone_desc in tones.items():
    print(f'\n{"=" * 50}')
    print(f'TONE: {tone_name}')
    print('=' * 50)
    prompt = f'{base_context}\n\n{task}\n\nTone: {tone_desc}\nInclude subject line.'
    print(ask(prompt, temperature=0.4))

print('\nREFLECTION: Which tone is right depends on context, relationship, and intent.')
print('AI gives you all 3 in seconds — you choose the right one.')
```

# SESSION 12 — AI for Business Reports & Documents

### Lab 12.1 — Phased Report Generation
**Task:** Generate a business report in 3 phases: Outline → Draft → Polish.

```python
# Lab 12.1 — 3-Phase Report Generation
TOPIC = 'Implementing a Hybrid Work Policy: Recommendations for 2025'
CONTEXT = '250-person tech company, 60% WFH-eligible roles, attrition at 18%'

# Phase 1: Outline
outline = ask(
    f'Senior HR consultant. Create a 5-section report outline.\n'
    f'Topic: {TOPIC}\nContext: {CONTEXT}\n'
    f'For each section: title + 2-line description of content. Numbered list only.',
    temperature=0.3
)
print('PHASE 1 — OUTLINE:'); print(outline)

# Phase 2: Executive Summary
exec_summary = ask(
    f'Senior HR consultant. Write a 150-word executive summary.\n'
    f'Topic: {TOPIC}\nContext: {CONTEXT}\n'
    f'Include: business case (attrition as driver), 3 recommendation previews, Q1 decision call.\n'
    f'2 paragraphs. Professional tone.',
    temperature=0.4
)
print('\nPHASE 2 — EXECUTIVE SUMMARY:'); show(exec_summary)

# Phase 3: Polish
polished = ask(
    f'Professional editor. Improve this executive summary:\n'
    f'1. Eliminate ALL passive voice\n2. Replace any jargon with plain English\n'
    f'3. Strengthen the opening sentence\n4. Make the call-to-action more specific\n\n'
    f'Output: improved version + list of 3 changes made.\n\n{exec_summary}',
    temperature=0.3
)
print('\nPHASE 3 — POLISHED:'); show(polished)
```

# SESSION 14 — AI for Research & Summarization
## Labs: Layered Summarization, Multi-Source Synthesis, SIFT Framework

### Lab 14.1 — Multi-Source Research Synthesis
**Task:** Synthesize insights across 3 sources — identify consensus and tensions.

```python
# Lab 14.1 — Multi-Source Synthesis
source_a = 'Gartner 2024: 65% enterprises will use GenAI by 2025. Top uses: content (72%), code (61%), CX (54%).'
source_b = 'McKinsey 2024: 79% companies use AI. Only 28% have governance. Productivity gains 15-20%.'
source_c = 'NASSCOM 2024: India AI investment +45% YoY. Talent shortage #1 barrier (68% CIOs). Need 1.2M professionals by 2026; have 420K.'

prompt = (
    f'Research analyst briefing C-suite at an Indian tech company.\n\n'
    f'Synthesize these 3 sources. Identify CROSS-SOURCE themes — do NOT summarize each separately.\n\n'
    f'Source A: {source_a}\nSource B: {source_b}\nSource C: {source_c}\n\n'
    f'Output:\n'
    f'1. CONSENSUS: What all 3 sources agree on (cite A/B/C)\n'
    f'2. KEY TENSION: Where sources disagree or reveal a gap\n'
    f'3. INDIA-SPECIFIC IMPLICATION: What this means for Indian companies\n'
    f'4. WATCH AREA: The one thing to monitor in the next 6 months\n\n'
    f'Max 250 words. Cite sources as [A], [B], [C].'
)

show(ask(prompt, temperature=0.4), 'Research Synthesis')
```

### Lab 14.2 — Layered Summarization
**Task:** Summarize at 3 levels of depth from the same source material.

```python
# Lab 14.2 — Layered Summarization
article = (
    'The Indian SaaS market has reached a significant inflection point in 2024, '
    'with total revenue crossing $14 billion. Growth is driven by three factors: '
    'increased enterprise adoption of cloud-native solutions, a maturing developer '
    'ecosystem producing globally competitive products, and favorable unit economics '
    'enabled by India\'s cost advantages in engineering talent. However, challenges '
    'persist. Customer acquisition costs have risen 35% YoY as competition intensifies. '
    'Net revenue retention for Indian SaaS companies averages 108%, compared to 120% '
    'for US counterparts, suggesting room for improvement in expansion revenue. '
    'The talent war is also reshaping the industry — senior engineering salaries have '
    'increased 40% since 2022, compressing margins for companies that haven\'t achieved '
    'scale. Analysts predict consolidation: 20-30% of current startups will be acquired '
    'or shut down within 18 months.'
)

layers = {
    'Elevator Pitch (30 words)': '30 words. 2 sentences max. For a CEO in a lift.',
    'Executive Brief (80 words)': '80 words. 3 key points. For a leadership team meeting.',
    'Detailed Analysis (200 words)': '200 words. 4 sections: Market State, Growth Drivers, Challenges, Outlook. For a strategy team.'
}

for layer_name, instruction in layers.items():
    prompt = f'Summarize this article at this depth: {instruction}\n\nArticle: {article}'
    print(f'\n{"=" * 50}')
    print(f'LAYER: {layer_name}')
    print('=' * 50)
    print(ask(prompt, temperature=0.3))

print('\nKEY: Same source, 3 audiences, 3 depth levels. AI handles all in seconds.')
```

### Lab 14.3 — SIFT Framework: Evaluating AI-Generated Research
**Task:** Apply the SIFT framework to assess the quality of AI research output.

```python
# Lab 14.3 — SIFT Framework for AI Research Evaluation
# First: generate a research-style output
research = ask(
    'Write a 200-word research summary about the impact of AI on employment in India in 2024. '
    'Include specific statistics, company examples, and predictions.',
    temperature=0.7
)
print('AI-GENERATED RESEARCH:'); print(research)

# Now: evaluate it with SIFT
sift_prompt = (
    f'Apply the SIFT framework to evaluate this AI-generated research output.\n\n'
    f'Output: {research}\n\n'
    f'SIFT Framework:\n'
    f'S — STOP: Before sharing, pause and evaluate critically.\n'
    f'I — INVESTIGATE the source: Are the statistics verifiable? Are company names real?\n'
    f'F — FIND better coverage: What claims need verification from primary sources?\n'
    f'T — TRACE claims: Can each specific number be traced to a real report?\n\n'
    f'For each element: evaluate this specific output. Flag any claims that:\n'
    f'- Cannot be verified (likely hallucinated)\n'
    f'- Need a primary source citation\n'
    f'- Are plausible but unconfirmed\n\n'
    f'Verdict: SAFE TO USE / NEEDS VERIFICATION / DO NOT USE WITHOUT CHECKING'
)

show(ask(sift_prompt, temperature=0.2), 'SIFT Evaluation')
print('\nLESSON: Never treat AI research output as fact. Always SIFT before sharing.')
```
