```python
# ============================================================
# GLOBAL SETUP
# ============================================================
import os, json, time, datetime
from IPython.display import Markdown, display
from openai import OpenAI

API_KEY = "sk-your-api-key-here"
client = OpenAI(api_key=API_KEY)

def ask(prompt, model='gpt-4o-mini', temperature=0.7, max_tokens=800, system=None):
    messages = []
    if system: messages.append({'role': 'system', 'content': system})
    messages.append({'role': 'user', 'content': prompt})
    r = client.chat.completions.create(model=model, messages=messages, temperature=temperature, max_tokens=max_tokens)
    return r.choices[0].message.content

def show(text, title=None):
    if title: display(Markdown(f'### {title}'))
    display(Markdown(text))

print('Setup complete.')
```

# SESSION 16 — AI for Marketing & Content Creation

### Lab 16.1 — Brand Voice Card
**Task:** Create a Brand Voice Card that anchors all AI-generated marketing content.

```python
# Lab 16.1 — Brand Voice Card Generator
COMPANY = 'A B2B SaaS company selling HR analytics software to mid-size enterprises'

prompt = (
    f'You are a senior brand strategist.\n\n'
    f'Company: {COMPANY}\n\n'
    f'Create a Brand Voice Card with these sections:\n'
    f'1. VOICE ATTRIBUTES (3 adjectives that define the brand voice)\n'
    f'2. VOICE DESCRIPTION (2 sentences explaining the overall tone)\n'
    f'3. WE SAY / WE DONT SAY (5 pairs of approved vs rejected phrases)\n'
    f'4. WRITING RULES (5 specific rules for all content)\n'
    f'5. SAMPLE PARAGRAPH (50 words demonstrating the voice perfectly)\n\n'
    f'This card will be pasted into every AI content prompt as a style anchor.'
)

brand_voice = ask(prompt, temperature=0.4)
show(brand_voice, 'Brand Voice Card')

# Now USE the brand voice to generate content
post = ask(
    f'Using this brand voice:\n{brand_voice}\n\n'
    f'Write a 200-word LinkedIn post about: Why HR teams that measure employee engagement weekly outperform those who survey annually.\n'
    f'Include: hook, 3 evidence points, engagement question. 5 hashtags.',
    temperature=0.6
)
show(post, 'LinkedIn Post with Brand Voice')
```

### Lab 16.2 — Content Repurposing Pipeline
**Task:** Take one blog post and repurpose it into 4 content formats.

```python
# Lab 16.2 — 1 Blog to 4 Formats
blog_summary = (
    'Why 80% of Customer Churn Is Predictable.\n'
    'Three signals appear 60-90 days before churn: login drops below 2x/week, '
    'support tickets shift to billing questions, NPS score drops below 7.\n'
    'Personal outreach retains 3x more at-risk customers than automated emails.\n'
    'A 15-min CSM call in the first 30 days of decline recovers 68% of at-risk accounts.'
)

formats = {
    'LinkedIn Post': 'Write a 180-220 word LinkedIn post. Hook + 3 insights + engagement question.',
    'Twitter/X Thread': 'Write a 6-tweet thread. Max 240 chars each. Hook -> 4 insights -> CTA.',
    'Email Subject Lines': 'Write 5 email subject lines: 2 curiosity-gap, 2 data-led, 1 benefit-first. Under 50 chars.',
    'Infographic Script': 'Write copy for a 5-panel infographic. Each panel: headline (5 words) + supporting stat or insight (10 words).'
}

for fmt_name, instruction in formats.items():
    prompt = f'Content strategist. Brand voice: direct, evidence-based.\n\nSource: {blog_summary}\n\n{instruction}'
    print(f'\n{"=" * 50}')
    print(f'FORMAT: {fmt_name}')
    print('=' * 50)
    print(ask(prompt, temperature=0.6))

print('\nRESULT: 1 blog -> 4 content pieces in under 2 minutes.')
```

### Lab 17.1 — Inclusive Job Description + Bias Audit
**Task:** Generate a JD, then audit it for bias — a 2-step responsible HR workflow.

```python
# Lab 17.1 — JD Generation + Bias Audit
notes = (
    'Role: Data Analyst, Marketing Analytics, mid-level.\n'
    'Must have: SQL, Python, Excel, presenting to non-technical teams.\n'
    'Nice to have: Tableau/Power BI.\n'
    'Responsibilities: dashboards, campaign analysis, A/B testing.'
)

# Step 1: Generate inclusive JD
jd = ask(
    f'Senior HRBP. Write an inclusive job description.\n\n'
    f'Notes: {notes}\n\n'
    f'Format: Title | About the Role (lead with IMPACT) | Responsibilities (5) | '
    f'Must-Have (separate from Nice-to-Have) | Benefits\n'
    f'Rules: No unnecessary degree requirements. Use gender-neutral language. '
    f'Focus on skills and outcomes, not years of experience.',
    temperature=0.4
)
print('STEP 1 — GENERATED JD:'); print(jd)

# Step 2: Bias audit
audit = ask(
    f'Responsible AI auditor specializing in HR bias.\n\n'
    f'Audit this job description for:\n'
    f'1. Masculine-coded language (list word + neutral replacement)\n'
    f'2. Ageist terms (flag + replace)\n'
    f'3. Over-qualification / unnecessary barriers\n'
    f'4. Culturally exclusive language\n\n'
    f'Format: | Issue | Type | Original | Recommended Fix |\n'
    f'If no issues found in a category, say "No issues detected."\n\n'
    f'JD: {jd}',
    temperature=0.1
)
print('\nSTEP 2 — BIAS AUDIT:'); show(audit)
```

# SESSION 19 — AI for Customer Support
## Labs: HEARD Framework, Response Templates, Chatbot Design

### Lab 19.1 — HEARD Framework Responses
**Task:** Generate empathy-first customer support responses using the HEARD framework.

```python
# Lab 19.1 — HEARD Framework Customer Responses
complaints = [
    {
        'complaint': 'Account downgraded to free with no notice. Lost all project data. Client demo in 4 hours.',
        'customer': 'Raghav Mehta',
        'context': '3-year premium subscriber. Account PRO-44821.'
    },
    {
        'complaint': 'Your chatbot gave me wrong billing information and I overpaid by Rs 4,500.',
        'customer': 'Sneha Patel',
        'context': 'First-time complainant. 6-month customer.'
    },
    {
        'complaint': 'This is the THIRD time my delivery has been late. I want a full refund and cancellation.',
        'customer': 'Arjun Krishnan',
        'context': '1-year customer. 2 previous late delivery complaints on file.'
    },
]

for c in complaints:
    prompt = (
        f'Senior customer success specialist. Apply the HEARD framework:\n'
        f'H = Hear (acknowledge the SPECIFIC problem)\n'
        f'E = Empathize (show you understand the impact)\n'
        f'A = Apologize (one sincere apology)\n'
        f'R = Resolve (specific action + timeline)\n'
        f'D = Diagnose (what caused this + how we prevent it)\n\n'
        f'Customer: {c["customer"]} | Context: {c["context"]}\n'
        f'Complaint: {c["complaint"]}\n\n'
        f'Under 180 words. Warm, urgent, accountable.\n'
        f'Do NOT use: "I understand your frustration", "Certainly!", "As per our policy"'
    )
    show(ask(prompt, temperature=0.4), f'Response to {c["customer"]}')
    print()
```
