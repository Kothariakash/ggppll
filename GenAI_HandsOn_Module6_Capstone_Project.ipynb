```python
# ============================================================
# GLOBAL SETUP
# ============================================================
import os, json, time, datetime
from IPython.display import Markdown, display
from openai import OpenAI

API_KEY = "sk-your-api-key-here"
client = OpenAI(api_key=API_KEY)

def ask(prompt, model='gpt-4o-mini', temperature=0.7, max_tokens=800, system=None):
    messages = []
    if system: messages.append({'role': 'system', 'content': system})
    messages.append({'role': 'user', 'content': prompt})
    r = client.chat.completions.create(model=model, messages=messages, temperature=temperature, max_tokens=max_tokens)
    return r.choices[0].message.content

def show(text, title=None):
    if title: display(Markdown(f'### {title}'))
    display(Markdown(text))

def craft_prompt(context, role, action, fmt, tone, constraints=None):
    p = f'CONTEXT: {context}\n\nROLE: {role}\n\nACTION: {action}\n\nFORMAT: {fmt}\n\nTONE: {tone}'
    if constraints: p += f'\n\nCONSTRAINTS: {constraints}'
    return p

print('Setup complete.')
```

### Lab 27.1 — Core Prompt: Build to Production Standard
**Task:** Write your most important prompt using the full CRAFT production template.

```python
# Lab 27.1 — Core Prompt (Production-Ready)
# CUSTOMIZE: This is YOUR capstone's most important prompt

core_prompt = craft_prompt(
    context=(
        'You are helping a marketing team write weekly campaign performance commentary. '
        'The audience is the VP of Sales and the CFO. They need clear, specific, '
        'actionable insights — not a data dump.'
    ),
    role=(
        'You are a senior marketing analyst with 10 years of B2B SaaS experience. '
        'You specialize in translating campaign metrics into business narratives.'
    ),
    action=(
        'Interpret the campaign data below and write the weekly performance narrative.\n'
        '1. HEADLINE: One sentence — the single most important finding this week\n'
        '2. TOP PERFORMER: Which campaign did best? Why? (with specific metric)\n'
        '3. UNDERPERFORMER: Which campaign missed? Root cause? Recommended action?\n'
        '4. OVERALL vs TARGET: Are we on track for the monthly goal?'
    ),
    fmt=(
        '4 sections with section headers. Professional prose — not bullet points. '
        'Under 200 words total. Active voice.'
    ),
    tone=(
        'Confident, specific, evidence-based. Written for executives who skim. '
        'Do NOT use: "performed well", "showed improvement", or any vague language.'
    ),
    constraints=(
        'Every claim must reference a specific number from the data. '
        'If data is insufficient to explain a variance, flag as [INVESTIGATE: ...]. '
        'Do not invent explanations for performance changes.'
    )
)

# Test with sample data
sample_data = (
    'Campaign: Google Search | Spend: 1.2L | Leads: 48 | Target: 50 | CPL: 2,500\n'
    'Campaign: LinkedIn Ads  | Spend: 0.8L | Leads: 12 | Target: 20 | CPL: 6,667\n'
    'Campaign: Email Nurture | Spend: 0.1L | Leads: 35 | Target: 30 | CPL: 286\n'
    'Campaign: Webinar       | Spend: 0.3L | Leads: 22 | Target: 25 | CPL: 1,364\n'
    'Monthly target: 200 leads. MTD: 117 leads (Week 2 of 4).'
)

full_prompt = core_prompt + f'\n\n---CAMPAIGN DATA---\n{sample_data}\n---END DATA---'
output = ask(full_prompt, temperature=0.3)
show(output, 'Core Prompt Output — Test Run 1')

# Quality Rubric Self-Score
rubric_prompt = (
    f'Score this prompt on the 6-dimension Quality Rubric (each 1-5):\n'
    f'1. Context | 2. Specificity of Action | 3. Format Specification | '
    f'4. Tone and Voice | 5. Constraints and Guardrails | 6. Testability\n\n'
    f'Markdown table with Score + Reason. Total /30.\n\n'
    f'Prompt:\n"{core_prompt}"'
)
show(ask(rubric_prompt, temperature=0.2), 'Quality Rubric Score')
```

# SESSION 28 — Capstone: Quality Assurance & Testing
## Labs: 5-Phase Testing, Prompt Fixing, Responsible AI Audit

### Lab 28.1 — Baseline and Stress Testing
**Task:** Run 10 systematic tests on your core prompt.

```python
# Lab 28.1 — Systematic Testing

# Baseline Tests (good inputs)
baseline_inputs = [
    'Google Search: 1.5L spend, 55 leads, target 50, CPL 2727\nLinkedIn: 1.0L, 18 leads, target 20, CPL 5556\nEmail: 0.1L, 40 leads, target 35, CPL 250\nMonthly target: 200. MTD: 113 (Week 2).',
    'Facebook Ads: 2.0L, 80 leads, target 60, CPL 2500\nGoogle Display: 0.5L, 15 leads, target 20, CPL 3333\nContent Syndication: 0.8L, 25 leads, target 30, CPL 3200\nWebinar: 0.2L, 30 leads, target 25, CPL 667\nMonthly target: 250. MTD: 150 (Week 3).',
    'SEO: 0L spend, 120 leads, target 100\nPaid Search: 3.0L, 90 leads, target 80, CPL 3333\nPartner Referral: 0L, 15 leads, target 20\nMonthly target: 300. MTD: 225 (Week 3).',
]

# Stress Tests (challenging inputs)
stress_inputs = [
    'Google: 1L, 5 leads, target 50. Thats it.',  # Minimal / incomplete
    'All campaigns performing at exactly target. No variance.',  # No story to tell
    'Campaign A: 0.5L, 100 leads, target 10, CPL 500\nCampaign B: 5.0L, 2 leads, target 50, CPL 250000',  # Extreme variance
]

test_log = []
all_inputs = [('Baseline', i, inp) for i, inp in enumerate(baseline_inputs, 1)] + \
             [('Stress', i, inp) for i, inp in enumerate(stress_inputs, 1)]

for phase, num, test_data in all_inputs:
    full = core_prompt + f'\n\n---CAMPAIGN DATA---\n{test_data}\n---END DATA---'
    result = ask(full, temperature=0.3)
    
    # Quick quality check
    has_numbers = any(c.isdigit() for c in result)
    under_limit = len(result.split()) <= 250
    status = 'PASS' if (has_numbers and under_limit) else 'CHECK'
    
    print(f'\n{"=" * 50}')
    print(f'TEST {phase}-{num} | Status: {status}')
    print(f'Input: {test_data[:80]}...')
    print(f'Output ({len(result.split())} words):')
    print(result[:300] + '...' if len(result) > 300 else result)
    
    test_log.append({'phase': phase, 'num': num, 'status': status, 'word_count': len(result.split())})

print(f'\n{"=" * 50}')
print(f'TEST SUMMARY: {sum(1 for t in test_log if t["status"]=="PASS")}/{len(test_log)} passed')
print('Review each CHECK result — what failed and why?')
```

### Lab 28.2 — Adversarial Testing
